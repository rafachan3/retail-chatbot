# Data Processing for Retail Chatbot

This notebook processes and cleans the retail product dataset for use in the chatbot recommendation engine.

## Key Processing Steps:
1. Load and concatenate multiple CSV files
2. Remove numerical columns (except article_id) 
3. Filter out redundant categorical columns
4. Remove missing values
5. Filter out kid/baby items for adult-focused recommendations
6. Export cleaned dataset for the recommendation engine

```python
import pandas as pd
```

```python
# Read and concatenate all CSV files using a loop
dfs = []
for i in range(1, 12):
    filename = f'articles_with_expanded_mood_style_part_{i}.csv'
    dfs.append(pd.read_csv(filename))

df = pd.concat(dfs, ignore_index=True)
print(len(df), "rows loaded from 11 files")
df.head()
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].value_counts().head(10))
    print("-" * 50)
```

```python
numerical_cols = df.select_dtypes(include=['number']).columns.to_list()
numerical_cols.remove('article_id')  # Exclude article_id if present
df[numerical_cols].describe()
```

## Column filtering

- We don't need the numerical values for the purpose of the project, with the exception of article_id
- Some of the categorical columns are redundant

```python
df.drop([*numerical_cols, 'index_group_name', 'perceived_colour_master_name', 'index_code'], axis=1, inplace=True)
```

```python
df.head()
```

## Missing values

```python
for col in df.columns:
    if df[col].isna().sum() > 0:
        print(f"Column '{col}' has {df[col].isna().sum()} missing values.")
```

```python
df.dropna(inplace=True)
print(f'The dataframe has {df.isna().sum().sum()} missing values.')
```

```python
print("Top 20 values of each feature:")
for col in df.columns:
    if col not in ['article_id']:
        print(f"{col}: {df[col].value_counts().head(20)}")
        if 'Unknown' in df[col].unique() or 'Undefined' in df[col].unique():
            print("Unknown values found:", len(df[df[col] == 'Unknown']))
```

```python
df.to_csv('articles_expanded_cleaned.csv', index=False)
```

## Remove items that include kid or baby in their department_name

```python
df = pd.read_csv('articles_expanded_cleaned.csv')
df.head()
```

```python
# Remove rows that contain 'kid' or 'baby' in department_name (case insensitive)
df = df[~df['department_name'].str.contains('kid|baby', case=False, na=False)]
print(f"Rows remaining after removing kid/baby items: {len(df)}")
```

```python
df.to_csv('articles_expanded_cleaned.csv', index=False)
print(f"Saved {len(df)} rows to articles_expanded_cleaned.csv (kid/baby items removed)")
```
